```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
import re
# Step 1. Load Pre-trained Model (DialoGPT)

model_name = "microsoft/DialoGPT-medium"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token
print("Model loaded successfully!\n")

# Step 2. Chatbot Initialization

chat_history_ids = None

print("Chatbot: Hello! I am your AI assistant. Type 'exit' to stop.")

while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        print("Chatbot: Goodbye!")
        break

    # Encode user input
    new_input_ids = tokenizer.encode(user_input + tokenizer.eos_token, return_tensors='pt')

    # Keep chat history
    bot_input_ids = torch.cat([chat_history_ids, new_input_ids], dim=-1) if chat_history_ids is not None else new_input_ids

    # Generate response
    chat_history_ids = model.generate(
        bot_input_ids,
        max_length=1000,
        pad_token_id=tokenizer.eos_token_id,
        do_sample=True,
        top_k=100,
        top_p=0.95,
        temperature=0.75
    )

    # Decode response
    response = tokenizer.decode(chat_history_ids[:, bot_input_ids.shape[-1]:][0], skip_special_tokens=True)

    print("Chatbot:", response)

# 3. Continuous Conversation Loop
while True:

    # Take user input
    user_input = input("User: ")

    # Exit condition
    if user_input.lower() in ["exit", "quit"]:
        print("Chatbot: Goodbye! Have a great day!")
        break

    # Check for predefined responses
    raw_user_input_for_debug = user_input # Keep original for debug if needed
    cleaned_user_input = user_input.lower().strip()
    # This addresses potential hidden characters not caught by strip/lower
    cleaned_user_input = re.sub(r'[^a-z0-9\s?]', '', cleaned_user_input)
    cleaned_user_input = cleaned_user_input.strip() # Re-strip after regex if new spaces appear

    cleaned_user_input = cleaned_user_input.encode('utf-8', 'ignore').decode('utf-8')
    # Explicit comparison for debugging specific cases
    if 'hello' in predefined_responses:
        expected_key = 'hello'

    if 'what is artificial intelligence?' in predefined_responses:
        expected_key = 'what is artificial intelligence?'
    if cleaned_user_input in predefined_responses:
        print("Chatbot:", predefined_responses[cleaned_user_input])
    else:
        # Encode user input
        new_input_ids = tokenizer.encode(user_input + tokenizer.eos_token, return_tensors='pt')

        # Append chat history
        bot_input_ids = torch.cat([chat_history_ids, new_input_ids], dim=-1) if chat_history_ids is not None else new_input_ids

        # Generate response
        chat_history_ids = model.generate(
            bot_input_ids,
            attention_mask=torch.ones(bot_input_ids.shape, dtype=torch.long),
            max_length=1000,
            pad_token_id=tokenizer.eos_token_id,
            do_sample=True,
            top_k=50,
            top_p=0.95,
            temperature=0.75
        )

        # Decode response
        response = tokenizer.decode(chat_history_ids[:, bot_input_ids.shape[-1]:][0], skip_special_tokens=True)
        print("Chatbot:", response)
```
